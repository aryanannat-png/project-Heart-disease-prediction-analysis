# Multivariate Analysis


Multivariate analysis examines the relationship among three or more variables simultaneously. It helps identify patterns, interactions, and associations that may not be visible through univariate or bivariate analysis.

## Systolic Pressure vs Diastolic Pressure by Heart Disease

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='systolic_bp',
    y='diastolic_bp',
    hue='disease',
    palette='Set1',
    s=80
)

plt.title('Systolic vs Diastolic Pressure by Heart Disease')
plt.xlabel('Systolic Pressure (mmHg)')
plt.ylabel('Diastolic Pressure (mmHg)')
plt.legend(title='Heart Disease')

plt.show()
```

### 💡 Insight

The scatter plot visualizes the relationship between systolic and diastolic blood pressure for individuals with and without heart disease.
Differences in the distribution of the two groups may indicate an association between blood pressure patterns and heart disease.
Clusters or overlapping regions help identify whether blood pressure measurements alone clearly distinguish heart disease status.

## Relationship Between Smoking, Alcohol Consumption, and Heart Disease
This analysis examines the relationship between smoking and alcohol consumption while considering heart disease status. A grouped count plot is used to compare the distribution of individuals with and without heart disease across different smoking and alcohol consumption categories.


```python
g = sns.catplot(
    data=df,
    x='smoking',
    hue='alcohol_consumption',
    col='disease',
    kind='count',
    height=5,
    aspect=0.9
)

g.fig.suptitle('Smoking vs Alcohol Consumption by Heart Disease', y=1.05)

plt.show()
```

### 💡 Insight
1. The analysis shows that individuals with heart disease have a higher number of smokers compared to individuals without heart disease.
2. Among individuals with heart disease, smoking is more common irrespective of alcohol consumption, indicating a stronger association between smoking and heart disease in this dataset.
3. In contrast, the non-heart disease group contains a higher proportion of non-smokers, suggesting that non-smoking individuals are less represented among heart disease cases.
4. Overall, the visualization indicates that smoking, particularly when combined with alcohol consumption, is more frequently observed among individuals with heart disease, highlighting these lifestyle factors as potential risk indicators.
